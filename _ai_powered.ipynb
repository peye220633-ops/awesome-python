````python
corrected_app_content = """
import streamlit as st
import yfinance as yf
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from ta.trend import MACD
from ta.momentum import RSIIndicator
import openai

# ตั้งค่าหน้าจอ
st.set_page_config(layout="wide", page_title="AI Technical Analysis Artifact")
st.title("📈 AI-Powered Interactive Technical Analysis")

# --- Sidebar Inputs ---
st.sidebar.header("Configuration")
ticker = st.sidebar.text_input("Enter Ticker (e.g., AAPL, BTC-USD)", value="AAPL")
time_period = st.sidebar.selectbox("Period", ["1mo", "3mo", "6mo", "1y", "2y"], index=2)
interval = st.sidebar.selectbox("Interval", ["60m", "1d", "1wk"], index=1)
api_key = st.sidebar.text_input("OpenAI API Key", type="password")

# --- Data Fetching & Calculation ---
@st.cache_data(ttl=3600)
def load_data(ticker, period, interval):
    df = yf.download(ticker, period=period, interval=interval)
    if df.empty:
        return df

    # คำนวณ Indicators
    df['MA20'] = df['Close'].rolling(window=20).mean()
    df['MA50'] = df['Close'].rolling(window=50).mean()
    df['RSI'] = RSIIndicator(close=df['Close']).rsi()
    macd = MACD(close=df['Close'])
    df['MACD'] = macd.macd()
    df['MACD_Signal'] = macd.macd_signal()
    return df

df = load_data(ticker, time_period, interval)

if df.empty:
    st.error("ไม่พบข้อมูล กรุณาตรวจสอบ Ticker อีกครั้ง")
else:
    # --- Interactive Chart (Plotly) ---
    fig = make_subplots(rows=3, cols=1, shared_xaxes=True,
                            vertical_spacing=0.05,
                                            row_width=[0.2, 0.2, 0.6])

    # Row 1: Candlestick & MAs
    fig.add_trace(go.Candlestick(x=df.index, open=df['Open'], high=df['High'], low=df['Low'], close=df['Close'], name="Price"), row=1, col=1)
    fig.add_trace(go.Scatter(x=df.index, y=df['MA20'], name='MA 20', line=dict(color='orange', width=1.5)), row=1, col=1)
    fig.add_trace(go.Scatter(x=df.index, y=df['MA50'], name='MA 50', line=dict(color='blue', width=1.5)), row=1, col=1)

    # Row 2: RSI
    fig.add_trace(go.Scatter(x=df.index, y=df['RSI'], name='RSI', line=dict(color='purple')), row=2, col=1)
    fig.add_hline(y=70, line_dash="dash", line_color="red", row=2, col=1)
    fig.add_hline(y=30, line_dash="dash", line_color="green", row=2, col=1)

    # Row 3: MACD
    fig.add_trace(go.Scatter(x=df.index, y=df['MACD'], name='MACD', line=dict(color='macro')), row=3, col=1)
    fig.add_trace(go.Scatter(x=df.index, y=df['MACD_Signal'], name='Signal', line=dict(color='red')), row=3, col=1)

    fig.update_layout(height=700, xaxis_rangeslider_visible=False, template="plotly_dark")
    st.plotly_chart(fig, use_container_width=True)

    # --- AI Analysis Section ---
    st.subheader("🤖 AI Market Insights")

    if not api_key:
        st.info("💡 ใส่ OpenAI API Key ที่ Sidebar เพื่อเปิดใช้งานระบบวิเคราะห์ด้วย AI")
    else:
        # เตรียมข้อมูลสรุปล่าสุดให้ AI
        latest = df.iloc[-1]
        prev = df.iloc[-2]

        prompt_template = (
            "Analyze the following technical data for {ticker_val} (Interval: {interval_val}):\n"
            "    - Current Price: {latest_close_formatted} (Previous: {prev_close_formatted})\n"
            "    - MA20: {ma20_val_formatted}, MA50: {ma50_val_formatted}\n"
            "    - RSI: {rsi_val_formatted}\n"
            "    - MACD: {macd_val_formatted}, Signal: {macd_signal_val_formatted}\n"
        )
        prompt = prompt_template.format(
            ticker_val=ticker,
            interval_val=interval,
            latest_close_formatted=f"{latest['Close']:.2f}",
            prev_close_formatted=f"{prev['Close']:.2f}",
            ma20_val_formatted=f"{latest['MA20']:.2f}",
            ma50_val_formatted=f"{latest['MA50']:.2f}",
            rsi_val_formatted=f"{latest['RSI']:.2f}",
            macd_val_formatted=f"{latest['MACD']:.2f}",
            macd_signal_val_formatted=f"{latest['MACD_Signal']:.2f}"
        )

        if st.button("Generate AI Analysis"):
            with st.spinner("AI กำลังวิเคราะห์กราฟและตัวชี้วัด..."):
                try:
                    openai.api_key = api_key
                    response = openai.ChatCompletion.create(
                        model="gpt-4o",
                        messages=[
                            {"role": "system", "content": "You are an expert financial analyst skilled in technical analysis."},
                            {"role": "user", "content": prompt}
                        ]
                    )
                    st.write(response.choices[0].message['content'])
                except Exception as e:
                    st.error(f"Error: {str(e)}")
"""

with open('app.py', 'w') as f:
    f.write(corrected_app_content)

print("app.py has been updated with the corrected content.")
````
